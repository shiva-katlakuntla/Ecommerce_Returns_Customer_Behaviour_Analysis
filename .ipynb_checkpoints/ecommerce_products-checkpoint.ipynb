## Cleaning Data

### Imputing `sub_category`

Recalculating the mode for `sub_category` grouped by `category` and filling NaN values.

```python
  products.replace(r'\N', np.nan, inplace=True)
print("Values '\N' replaced with NaN:")
print(products.isnull().sum())
```

```python
# Recalculate mode for 'sub_category' after replacing '\N' with NaN
# Now that '\N' has been replaced by np.nan, this will correctly find modes of valid categories
sub_category_mode = products.groupby("category")["sub_category"].apply(lambda x: x.mode()[0] if not x.mode().empty else np.nan)

# Fill NaN values in 'sub_category' with the calculated mode
products['sub_category'] = products['sub_category'].fillna(products['category'].map(sub_category_mode))
```

### Imputing `return_eligible`

Recalculating the mode for `return_eligible` grouped by `category` and `sub_category` and filling NaN values.

```python
# Recalculate mode for 'return_eligible' after replacing '\N' with NaN
# Now that '\N' has been replaced by np.nan, this will correctly find modes of valid return eligibilities
eligible_mode = products.groupby(["category","sub_category"])["return_eligible"].apply(lambda x: x.mode()[0] if not x.mode().empty else np.nan)

# Fill NaN values in 'return_eligible' using the re-calculated mode
products['return_eligible'] = products.apply(lambda row: row['return_eligible'] if pd.notna(row['return_eligible']) else eligible_mode.get((row['category'], row['sub_category']), np.nan), axis=1)
```

```python
# Final check for missing values after imputation
print("Missing values after imputation:")
print(products.isnull().sum())
```

### Cleaning `product_rating`

Converting `product_rating` to numeric and capping values between 0 and 5.

```python
# Convert 'product_rating' to numeric, coercing errors to NaN
products['product_rating'] = pd.to_numeric(products['product_rating'], errors='coerce')

# Fill any new NaNs created by coercion with the median rating
products['product_rating'].fillna(products['product_rating'].median(), inplace=True)

# Cap product_rating values between 0 and 5
products['product_rating'] = products['product_rating'].clip(lower=0, upper=5)
```
