CUSTOMERS TABLE


```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
customers=pd.read_csv("C:/Users/Shiva/Downloads/ecommerce_raw/customers.csv")
customers
```

```python
customers["gender"]=customers["gender"].astype(str).str.lower()
```

```python
customers['gender'] = customers['gender'].replace({
    'm': 'Male',
    'male': 'Male',
    'f': 'Female',
    'female': 'Female',
    'other': 'Other'
})
```

```python
customers.loc[customers['age'] < 0, 'age'] = 0
```

```python
customers['age']
```

```python
customers[customers['age'].isnull()]
```

```python
customers = customers[customers['customer_id'] != 1373]
```

```python
customers["city"]=customers["city"].astype(str).str.title()
customers["state"]=customers["state"].astype(str).str.title()
customers["loyalty_tier"]=customers["loyalty_tier"].astype(str).str.title()
customers["customer_status"]=customers["customer_status"].astype(str).str.title()
```

```python
customers
```

```python
state_mode = customers.groupby("city")["state"].apply(lambda x: x.mode()[0])
customers['state'] = customers.groupby('city')['state'].transform(lambda x: x.fillna(x.mode()[0]))
```

```python
loyalty_tier_mode = customers['loyalty_tier'].mode()[0]
print(f"The mode of 'loyalty_tier' is: {loyalty_tier_mode}")
```

```python
customers['loyalty_tier'] = customers['loyalty_tier'].replace("Nan",loyalty_tier_mode)
```

```python
annual_spend_median = customers.groupby("loyalty_tier")['annual_spend'].apply(lambda x:x.median())
print(f"The median of 'annual_spend' is: {annual_spend_median}")
```

```python
customers['annual_spend'] = customers.groupby("loyalty_tier")['annual_spend'].transform(lambda x: x.fillna(x.median()))
```

```python
customers.isnull().sum()
```

```python
customers["age"].value_counts()
```

```python
customers['age'].describe()
```

```python
Q1 = customers['age'].quantile(0.25)
Q3 = customers['age'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("lower bound:",lower_bound)
print("upper bound:", upper_bound)

age_outliers = customers[
    (customers['age'] < lower_bound) |
    (customers['age'] > upper_bound)
]
print( "count no of ouliers:",age_outliers.shape[0])
age_outliers
```

### City-wise Customers

```python
city_customers = customers['city'].value_counts()

city_customers
```

```python
city_customers.head(10).plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Top Cities by Customer Count')
plt.xlabel('City')
plt.ylabel('Customers')

plt.show()
```

### Loyalty Membership Analysis

```python
customers['loyalty_tier'].value_counts()
```

```python
customers['loyalty_tier'].value_counts().plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Loyalty Tier Distribution')
plt.xlabel('Loyalty Tier')
plt.ylabel('Customers')

plt.show()
```

### Which Age Group Purchases More?

```python
customers['age_group'] = pd.cut(
    customers['age'],
    bins=[0,18,25,35,45,60,100],
    labels=[
        '0-18',
        '19-25',
        '26-35',
        '36-45',
        '46-60',
        '60+'
    ]
)

age_group_spend = customers.groupby(
    'age_group'
)['annual_spend'].sum().sort_values(ascending=False)

age_group_spend
```

```python
age_group_spend.plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Total Spend by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Annual Spend')

plt.show()
```

### Which Cities Generate Highest Revenue?

```python
city_revenue = customers.groupby(
    'city'
)['annual_spend'].sum().sort_values(ascending=False)

city_revenue.head(10)
```

```python
city_revenue.head(10).plot(
    kind='bar',
    figsize=(12,5)
)

plt.title('Top Revenue Generating Cities')
plt.xlabel('City')
plt.ylabel('Revenue')

plt.show()
```

### Loyal vs Non-Loyal Customers

```python
customers['customer_type'] = np.where(
    customers['loyalty_tier'].isin(
        ['Silver','Gold','Platinum']
    ),
    'Loyal',
    'Non-Loyal'
)

print("compare customers:",customers['customer_type'].value_counts())

loyal_revenue = customers.groupby(
    'customer_type'
)['annual_spend'].sum()

print(loyal_revenue)
```

```python
loyal_revenue.plot(
    kind='bar',
    figsize=(6,5)
)

plt.title('Loyal vs Non-Loyal Revenue')
plt.xlabel('Customer Type')
plt.ylabel('Revenue')

plt.show()
```
